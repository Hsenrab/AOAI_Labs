## Create Azure AI Foundry Workspace
1. Navigate to Azure AI Foundry: https://ai.azure.com
2. Sign in with your Azure credentials
3. Create a new workspace by clicking **"+ New workspace"**
   - Choose a workspace name
   - Select your Azure subscription
   - Create or select a resource group
   - Choose a region that supports Azure OpenAI
   - The workspace will automatically create associated storage and other required services
4. Note down your workspace details for configuration

## Deploy Azure OpenAI Models
1. In your Azure AI Foundry workspace at https://ai.azure.com
2. Navigate to **Models + endpoints** or **Deployments**
3. Click **+ Deploy model** and deploy the following models:
   - **GPT-4o** (for chat completions)
   - **text-embedding-ada-002** (for embeddings)
4. Note the deployment names
5. Ensure you have sufficient quota for these models

**Note**: Azure AI Foundry provides streamlined model deployment with built-in monitoring and management capabilities.

## Create .env file for Azure AI Foundry
Create a `.env` file in the Foundry101 directory with the following structure:

```env
# Azure AI Foundry Configuration
AZURE_SUBSCRIPTION_ID="your-subscription-id"
AZURE_RESOURCE_GROUP="your-resource-group"
AZURE_AI_WORKSPACE_NAME="your-foundry-workspace-name"

# Azure AI Foundry API (unified endpoint)
FOUNDRY_API_ENDPOINT="your-foundry-endpoint"
FOUNDRY_API_KEY="your-foundry-key"
OPENAI_API_VERSION="2024-08-01-preview"
GPT4O_DEPLOYMENT_NAME="gpt-4o"
EMBEDDINGS_DEPLOYMENT_NAME="text-embedding-ada-002"
```

## Getting the Configuration Values

### Azure AI Foundry Configuration
1. **Subscription ID**: Found in Azure portal → Subscriptions
2. **Resource Group**: The resource group containing your Azure AI Foundry workspace
3. **Workspace Name**: Your Azure AI Foundry workspace name (found at ai.azure.com)

### Azure AI Foundry API
1. Navigate to your Azure AI Foundry workspace at https://ai.azure.com
2. Go to **Settings** or **API Keys** section
3. Find your **Foundry endpoint** and **API key**
4. Note the deployment names you created in **Deployments**

**Where to find values:**
- **Foundry Endpoint**: In workspace settings at ai.azure.com
- **API Key**: In workspace API keys section
- **Deployment Names**: The names you gave when deploying models
